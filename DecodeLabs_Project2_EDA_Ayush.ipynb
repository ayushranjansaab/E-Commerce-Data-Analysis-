```python
plt.figure(figsize=(10,5))

sns.boxplot(x=df['TotalPrice'])

plt.title('Boxplot of Total Price')
plt.xlabel('Total Price')

plt.show()
```

```python
Q1 = df['TotalPrice'].quantile(0.25)
Q3 = df['TotalPrice'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = df[
    (df['TotalPrice'] < lower_bound) |
    (df['TotalPrice'] > upper_bound)
]

print("Number of Outliers:", len(outliers))
```

```python
numeric_df = df[['Quantity', 'UnitPrice', 'ItemsInCart', 'TotalPrice']]

correlation = numeric_df.corr()

correlation
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix')

plt.show()
```

### Correlation Analysis

The correlation analysis shows a strong positive relationship between UnitPrice and TotalPrice, with a correlation coefficient of 0.72. Quantity also has a moderately strong positive relationship with TotalPrice, with a correlation of 0.62. Quantity and ItemsInCart have a positive correlation of 0.65. In contrast, Quantity and UnitPrice have almost no linear relationship, with a correlation of only 0.01. Overall, UnitPrice and Quantity appear to be important factors associated with TotalPrice.

```python
Q1 = df['TotalPrice'].quantile(0.25)
Q3 = df['TotalPrice'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

### Outlier Analysis

The boxplot and IQR method were used to identify potential outliers in TotalPrice. The first quartile (Q1) is ₹410.52 and the third quartile (Q3) is ₹1,578.475, resulting in an IQR of ₹1,167.955. The calculated upper bound is approximately ₹3,330.41. A total of 8 observations were identified as potential outliers because their TotalPrice values exceeded this upper bound. These orders have unusually high transaction values, ranging from approximately ₹3,334 to ₹3,456. These observations were retained for analysis because they represent high-value orders rather than confirmed data errors.
